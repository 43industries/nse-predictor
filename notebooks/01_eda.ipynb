```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..').resolve()
sys.path.insert(0, str(ROOT / 'src'))

prices_path = ROOT / 'data' / 'processed' / 'prices.parquet'
if not prices_path.exists():
    print('Run scripts/run_ingest.py first')
else:
    prices = pd.read_parquet(prices_path)
    print(prices.shape, prices['trade_date'].min(), prices['trade_date'].max())
    print('tickers', prices['stock_code'].nunique())
```

```python
if prices_path.exists():
    liq = (
        prices.groupby('stock_code')
        .agg(avg_vol=('volume','mean'), zero_days=('zero_volume','sum'), n=('trade_date','count'))
        .sort_values('avg_vol', ascending=False)
    )
    display(liq.head(20))
    liq.head(15)['avg_vol'].plot(kind='bar', title='Avg daily volume — top 15')
    plt.tight_layout()
    plt.show()
```

```python
if prices_path.exists():
    for code in ['SCOM','EQTY','KCB','NASI']:
        s = prices[prices.stock_code==code].sort_values('trade_date')
        if s.empty:
            continue
        plt.figure(figsize=(10,3))
        plt.plot(s.trade_date, s.price)
        plt.title(code)
        plt.tight_layout()
        plt.show()
```
